```python
# -*- coding: utf-8 -*-
"""
Cantidad de datos por etiqueta (distribución / desbalance de clases).
Reporta, por cada uno de los 13 labels CheXpert (Support Devices excluido):
positivos, negativos, prevalencia, ratio neg/pos y el pos_weight con factor 0.75.
Codificación U-zeros (NaN y -1 -> 0), consistente con los notebooks.
"""
 
import pandas as pd
import numpy as np
 
# --- Configuración ---
RUTA = r"C:\Users\trodr\Documents\proyecto-torax-v2.0\01-dataset\dataset-completo-merge\df_subset_2994_valido.csv"
FACTOR_POS_WEIGHT = 0.75  # valor de producción validado
 
LABELS = [
    "Atelectasis", "Cardiomegaly", "Consolidation", "Edema",
    "Enlarged Cardiomediastinum", "Fracture", "Lung Lesion", "Lung Opacity",
    "No Finding", "Pleural Effusion", "Pleural Other", "Pneumonia", "Pneumothorax",
]
 
# --- Carga ---
df = pd.read_csv(RUTA)
 
faltantes = [l for l in LABELS if l not in df.columns]
if faltantes:
    raise ValueError(f"Estas etiquetas no están en el CSV (revisa nombres): {faltantes}")
 
# --- U-zeros: NaN y -1 -> 0 ---
etiquetas = df[LABELS].replace(-1, 0).fillna(0).astype(int)
n_total = len(etiquetas)
pos = etiquetas.sum()
 
# --- Tabla resumen ---
resumen = pd.DataFrame({"Positivos": pos, "Negativos": n_total - pos})
resumen["Prevalencia (%)"] = (resumen["Positivos"] / n_total * 100).round(2)
resumen["Ratio neg/pos"] = (resumen["Negativos"] / resumen["Positivos"].replace(0, np.nan)).round(2)
resumen["pos_weight (x0.75)"] = (resumen["Ratio neg/pos"] * FACTOR_POS_WEIGHT).round(2)
resumen.index.name = "Etiqueta"
resumen = resumen.sort_values("Positivos", ascending=False)  # cambia a ascending=True para ver primero los raros
 
print(f"Total de imágenes: {n_total}\n")
print(resumen.to_string())
 
# En Jupyter, para verla con formato:
# display(resumen)
 
# Opcional: exportar a Excel
# resumen.to_excel("conteo_por_etiqueta.xlsx")
```

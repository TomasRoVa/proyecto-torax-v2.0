```python
import pandas as pd

# Cargar archivos
labels   = pd.read_csv(r"C:\Users\trodr\Documents\proyecto-torax-v2.0\01-dataset\mimic-cxr-2.1.0-test-set-labeled.csv")
metadata = pd.read_csv(r"C:\Users\trodr\Documents\proyecto-torax-v2.0\01-dataset\mimic-cxr-2.0.0-metadata.csv")

# union: dicom_id, subject_id, study_id (ya lo tenías)
union = metadata[['dicom_id', 'subject_id', 'study_id']].copy()

# Merge labels (solo tienen study_id) con union (que tiene dicom_id)
df = union.merge(labels, on='study_id', how='inner')

# Construir full_path
BASE_DIR = r"C:\Users\trodr\Documents\proyecto-torax\01-dataset\archive\official_data_iccv_final"
p_prefix = union['subject_id'].astype(str).str[:2]

df['full_path'] = (
    BASE_DIR + "\\files"
    + "\\p" + df['subject_id'].astype(str).str[:2]
    + "\\p" + df['subject_id'].astype(str)
    + "\\s" + df['study_id'].astype(str)
    + "\\" + df['dicom_id'].astype(str) + ".jpg"
)

print(df.shape)
print(df[['dicom_id', 'subject_id', 'study_id', 'full_path']].head())
```

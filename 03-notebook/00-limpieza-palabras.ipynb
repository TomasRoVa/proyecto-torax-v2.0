```python
import os
import pandas as pd

BASE_DIR = r"C:\Users\trodr\Documents\proyecto-torax-v2.0\01-dataset\dataset_prueba\subset_pa_ap"
labels = pd.read_csv(r"C:\Users\trodr\Documents\proyecto-torax-v2.0\01-dataset\mimic-cxr-2.1.0-test-set-labeled.csv")

# Study IDs en disco (de los nombres de carpeta)
study_ids_disco = set(int(c[1:]) for c in os.listdir(BASE_DIR) if c.startswith('s'))
print(f"Study IDs en disco: {len(study_ids_disco)}")

# Study IDs en labels
study_ids_labels = set(labels['study_id'])
print(f"Study IDs en test-set-labeled: {len(study_ids_labels)}")

# ¿Hay algún match?
inter = study_ids_disco & study_ids_labels
print(f"Intersección: {len(inter)}")

# Muestrear 5 de cada lado para inspección
print(f"\n5 ejemplos de disco: {list(study_ids_disco)[:5]}")
print(f"5 ejemplos de labels: {list(study_ids_labels)[:5]}")

# Verificar dtypes
print(f"\ndtype de labels['study_id']: {labels['study_id'].dtype}")
```

```python
import os
import pandas as pd

# Cargar archivos — ahora usando chexpert
labels   = pd.read_csv(r"C:\Users\trodr\Documents\proyecto-torax-v2.0\01-dataset\mimic-cxr-2.0.0-chexpert.csv")
metadata = pd.read_csv(r"C:\Users\trodr\Documents\proyecto-torax-v2.0\01-dataset\mimic-cxr-2.0.0-metadata.csv")

# Union: dicom_id, subject_id, study_id
union = metadata[['dicom_id', 'subject_id', 'study_id']].copy()

# Merge por subject_id + study_id (chexpert tiene ambos)
df = union.merge(labels, on=['subject_id', 'study_id'], how='inner')

# BASE_DIR del subset descargado
BASE_DIR = r"C:\Users\trodr\Documents\proyecto-torax-v2.0\01-dataset\dataset_prueba\subset_pa_ap"

# Construir full_path: s{study_id}/{dicom_id}.jpg
df['full_path'] = (
    BASE_DIR
    + "\\s" + df['study_id'].astype(str)
    + "\\" + df['dicom_id'].astype(str) + ".jpg"
)

# Filtrar a los que existen físicamente
df['exists'] = df['full_path'].apply(os.path.exists)
df_subset = df[df['exists']].drop(columns='exists').reset_index(drop=True)

print(f"Total en df completo:  {len(df)}")
print(f"Total en subset local: {len(df_subset)}")
print(df_subset[['dicom_id', 'subject_id', 'study_id', 'full_path']].head())
```
